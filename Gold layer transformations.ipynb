```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
%sql
LIST 'abfss://silver@adlshealthcaredata.dfs.core.windows.net/';
```

```python
df_gold = spark.read.format("parquet") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("abfss://silver@adlshealthcaredata.dfs.core.windows.net/healthcare_dataset")

df_gold.display()
```

```python
df_gold = df_gold.withColumn("Date of Admission", to_date(col("Date of Admission"), "MM/dd/yyyy")) \
    .withColumn("Discharge Date", to_date(col("Discharge Date"), "MM/dd/yyyy"))

df_gold.display()
```

```python
from pyspark.sql.functions import datediff, col

df_gold = df_gold.withColumn(
    "Length_of_Stay_Days",
    datediff(col("Discharge Date"), col("Date of Admission"))
)
df_gold.display()
```
